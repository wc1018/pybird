```python
from pybird.greenfunction import GreenFunction
from pybird.correlator import Correlator

import os, sys
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from classy import Class

dk, ds = 0.01, 5
kd = np.arange(0.005, 0.3, dk)
sd = np.arange(10, 200, ds)
z = 0.52
a = 1/(1+z)
```

```python
Y1 = 0.
G1t = 3/7.
V12t = 1/7.
G1 = 1.
```

```python
G = GreenFunction(Omega0_m=0.3,fluid_equation_of_state = 'w0wa',
                  EoS_dict={'w0':-0.95,'wa':0.05},
                 EFTDE = False)
print(G.Y(a=a)-Y1,
      G.mG1t(a=a)/G1t,
      G.mV12t(a=a)/V12t,
      G.G(a=a)/G1
     )
```

```python
G = GreenFunction(Omega0_m=0.3,fluid_equation_of_state = 'w0wa',
                  EoS_dict={'w0':-0.75,'wa':-1},
                 EFTDE = False)
print(G.Y(a=a)-Y1,
      G.mG1t(a=a)/G1t,
      G.mV12t(a=a)/V12t,
      G.G(a=a)/G1
     )
```
